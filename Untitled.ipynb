```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

## Data Wrangling

```python
data_1 = pd.read_csv(r'car data.csv')
data_1.head(3)
```

```python
data_1['Selling_Price'] = data_1['Selling_Price'] * 100000
```

```python
data_2 = pd.read_csv(r'CAR DETAILS FROM CAR DEKHO.csv')
data_2.head(3)
```

```python
data_3 = pd.read_csv(r'Car details v3.csv')
data_3.head(3)
```

```python
data_4 = pd.read_csv(r'car details v4.csv')
data_4.head(3)
```

```python
data_4['name'] = data_4['Make'].astype(str) + ' ' + data_4['Model'].astype(str)
data_4.drop(columns = ['Make', 'Model'], inplace = True)
```

```python
data_1 = data_1.rename(
    columns = {
        'Car_Name' : 'name',
        'Year' : 'year',
        'Selling_Price' : 'selling_price',
        'Kms_Driven' : 'km_driven',
        'Fuel_Type' : 'fuel',
        'Seller_Type' : 'seller_type',
        'Transmission' : 'transmission',
        'Owner' : 'owner'
    }
).drop(columns = ['Present_Price'])
```

```python
data_4 = data_4.rename(
    columns = {
        'Year' : 'year',
        'Price' : 'selling_price',
        'Kilometer' : 'km_driven',
        'Fuel Type' : 'fuel',
        'Seller Type' : 'seller_type',
        'Transmission' : 'transmission',
        'Owner' : 'owner'
    }
)
```

```python
data_3 = data_3.rename(
    columns = {
        'year': 'year',
        'selling_price': 'selling_price',
        'km_driven': 'km_driven',
        'fuel': 'fuel',
        'seller_type': 'seller_type',
        'transmission': 'transmission',
        'owner': 'owner',
    }
)
```

```python
data_4 = data_4.rename(
    columns = {
        'Price': 'selling_price',
        'Year': 'year',
        'Kilometer': 'km_driven',
        'Fuel Type': 'fuel',
        'Seller Type': 'seller_type',
        'Transmission': 'transmission',
        'Owner': 'owner',
        'Engine': 'engine',
        'Max Power': 'max_power',
        'Max Torque': 'torque',
        'Seating Capacity': 'seats',
    }
)
```

```python
df = pd.concat([data_1, data_2, data_3, data_4], ignore_index = True)
```

```python
df.drop_duplicates(inplace = True, keep = 'first')
```

```python
df.isnull().sum()
```

```python
nan = pd.DataFrame(round(df.isnull().sum() * 100 / df.shape[0]), columns = ['NaN(%)'])
nan['D-type'] = df.dtypes.values
```

```python
nan
```

```python
df['Color'] = df['Color'].fillna('Others')
```

```python
str_less_50 = ['mileage', 'engine', 'max_power', 'torque']
num_less_50 = ['seats']
str_gre_50 = ['Location', 'Drivetrain']
num_gre_50 = ['Length', 'Width', 'Height', 'Fuel Tank Capacity']
```

```python
for col in str_less_50:
    df[col] = df[col].fillna(df[col].mode()[0])
    print(col, 'Done')
```

```python
for col in num_less_50:
    df[col] = df[col].fillna(df[col].median())
    print(col, 'Done')
```

```python
for col in str_gre_50:
    df[col] = df[col].fillna('Unknown')
    print(col, 'Done')
```

```python
for col in num_gre_50:
    df[col] = df[col].fillna(0.0)
    print(col, 'Done')
```

```python
df.isnull().sum()
```

```python
df['fuel'].unique()
```

```python
df['fuel'] = df['fuel'].replace({
    'CNG + CNG' : 'CNG'
})
```

```python
df['owner'] = df['owner'].replace({
    'First Owner' : 1,
    'Second Owner' : 2,
    'Fourth & Above Owner' : 4,
    'Third Owner' : 3,
    'Test Drive Car' : 0,
    'First' : 1,
    'Second' : 2,
    'Third' : 3,
    'Fourth' : 4,
    'UnRegistered Car' : 0,
    '4 or More' : 4
})
```

```python
df['mileage'] = df['mileage'].str.replace('kmpl', '', regex = False).str.strip()
df['mileage'] = df['mileage'].str.replace('km/kg', '', regex = False).str.strip()
```

```python
for i in df['mileage'].sample(500).unique():
    print(i)
```

```python
df['mileage'] = df['mileage'].astype(float)
```
